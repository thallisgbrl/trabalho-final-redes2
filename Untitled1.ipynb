```python
import numpy as np

def texto_para_bits(texto):
    bits = []
    for char in texto:
        binario = format(ord(char), '08b')
        bits.extend([int(b) for b in binario])
    return np.array(bits)

def bits_para_texto(array_bits):
    texto = ""
    for i in range(0, len(array_bits), 8):
        byte = array_bits[i:i+8]
        if len(byte) == 8:
            byte_str = ''.join(str(b) for b in byte)
            texto += chr(int(byte_str, 2))
    return texto

if __name__ == "__main__":
    mensagem_original = "Redes2_UNISINOS_2026"
    bits = texto_para_bits(mensagem_original)
    mensagem_recuperada = bits_para_texto(bits)

    print(f"Mensagem original: {mensagem_original}")
    print(f"Total de bits gerados: {len(bits)}")
    print(f"Mensagem recuperada: {mensagem_recuperada}")
    print(f"Sucesso? {mensagem_original == mensagem_recuperada}")
```

```python
import numpy as np

def encode_manchester(bits):
    sinal = np.zeros(2 * len(bits))
    for i, bit in enumerate(bits):
        if bit == 1:
            sinal[2*i:2*i+2] = [1, -1]
        else:
            sinal[2*i:2*i+2] = [-1, 1]
    return sinal

def decode_manchester(sinal):
    bits = []
    for i in range(0, len(sinal), 2):
        if sinal[i] > sinal[i+1]:
            bits.append(1)
        else:
            bits.append(0)
    return np.array(bits)

def encode_ami(bits):
    sinal = np.zeros(len(bits))
    polaridade = 1
    for i, bit in enumerate(bits):
        if bit == 1:
            sinal[i] = polaridade
            polaridade *= -1
        else:
            sinal[i] = 0
    return sinal

def decode_ami(sinal):
    bits = np.zeros(len(sinal), dtype=int)
    for i, valor in enumerate(sinal):
        if abs(valor) > 0.5:
            bits[i] = 1
        else:
            bits[i] = 0
    return bits

if __name__ == "__main__":
    bits_teste = np.array([1, 0, 1, 1, 0, 0, 1, 0])

    sinal_manchester = encode_manchester(bits_teste)
    bits_rec_manchester = decode_manchester(sinal_manchester)

    sinal_ami = encode_ami(bits_teste)
    bits_rec_ami = decode_ami(sinal_ami)

    print("Bits originais:    ", bits_teste)
    print("Sinal Manchester:  ", sinal_manchester)
    print("Recup. Manchester: ", bits_rec_manchester)
    print("Sinal AMI:         ", sinal_ami)
    print("Recup. AMI:        ", bits_rec_ami)
    print("Sucesso Manchester?", np.array_equal(bits_teste, bits_rec_manchester))
    print("Sucesso AMI?       ", np.array_equal(bits_teste, bits_rec_ami))
```

```python
import numpy as np

def mod_bpsk(bits):
    return np.where(bits == 1, 1 + 0j, -1 + 0j)

def demod_bpsk(symbols):
    return np.where(np.real(symbols) > 0, 1, 0)

def mod_qpsk(bits):
    simbolos = []
    for i in range(0, len(bits), 2):
        b1, b2 = bits[i], bits[i+1]
        re = 1 / np.sqrt(2) if b1 == 1 else -1 / np.sqrt(2)
        im = 1 / np.sqrt(2) if b2 == 1 else -1 / np.sqrt(2)
        simbolos.append(re + 1j*im)
    return np.array(simbolos)

def demod_qpsk(symbols):
    bits = []
    for s in symbols:
        bits.append(1 if np.real(s) > 0 else 0)
        bits.append(1 if np.imag(s) > 0 else 0)
    return np.array(bits)

def mod_16qam(bits):
    mapping = {(0,0): -3, (0,1): -1, (1,1): 1, (1,0): 3}
    simbolos = []
    if len(bits) % 4 != 0:
        bits = np.append(bits, np.zeros(4 - (len(bits) % 4), dtype=int))
    for i in range(0, len(bits), 4):
        re = mapping[(bits[i], bits[i+1])]
        im = mapping[(bits[i+2], bits[i+3])]
        simbolos.append((re + 1j*im) / np.sqrt(10))
    return np.array(simbolos)

def demod_16qam(symbols):
    bits = []
    symbols_norm = symbols * np.sqrt(10)
    for s in symbols_norm:
        re, im = np.real(s), np.imag(s)

        if re > 2:   b1, b2 = 1, 0
        elif re > 0: b1, b2 = 1, 1
        elif re > -2:b1, b2 = 0, 1
        else:        b1, b2 = 0, 0

        if im > 2:   b3, b4 = 1, 0
        elif im > 0: b3, b4 = 1, 1
        elif im > -2:b3, b4 = 0, 1
        else:        b3, b4 = 0, 0

        bits.extend([b1, b2, b3, b4])
    return np.array(bits)

if __name__ == "__main__":
    bits_teste = np.array([1, 0, 1, 1, 0, 0, 1, 0, 1, 1, 1, 1, 0, 0, 0, 0])

    sym_bpsk = mod_bpsk(bits_teste)
    rec_bpsk = demod_bpsk(sym_bpsk)

    sym_qpsk = mod_qpsk(bits_teste)
    rec_qpsk = demod_qpsk(sym_qpsk)

    sym_16qam = mod_16qam(bits_teste)
    rec_16qam = demod_16qam(sym_16qam)

    print("Bits originais: ", bits_teste)
    print("Sucesso BPSK?   ", np.array_equal(bits_teste, rec_bpsk))
    print("Sucesso QPSK?   ", np.array_equal(bits_teste, rec_qpsk))
    print("Sucesso 16-QAM? ", np.array_equal(bits_teste, rec_16qam))
```

```python
import numpy as np

def canal_awgn(sinal, snr_db):
    snr_linear = 10 ** (snr_db / 10)
    potencia_sinal = np.mean(np.abs(sinal) ** 2)
    potencia_ruido = potencia_sinal / snr_linear
    ruido = np.sqrt(potencia_ruido / 2) * (np.random.randn(len(sinal)) + 1j * np.random.randn(len(sinal)))
    return sinal + ruido

def canal_rajada(sinal, snr_db, prob_rajada=0.05, fator_rajada=10):
    sinal_ruidoso = canal_awgn(sinal, snr_db)
    mascara_rajada = np.random.rand(len(sinal)) < prob_rajada
    potencia_sinal = np.mean(np.abs(sinal) ** 2)
    potencia_ruido = potencia_sinal / (10 ** (snr_db / 10))
    ruido_rajada = np.sqrt((potencia_ruido * fator_rajada) / 2) * (np.random.randn(len(sinal)) + 1j * np.random.randn(len(sinal)))
    sinal_ruidoso[mascara_rajada] += ruido_rajada[mascara_rajada]
    return sinal_ruidoso

if __name__ == "__main__":
    bits_teste = np.array([1, 0, 1, 1, 0, 0, 1, 0, 1, 1, 1, 1, 0, 0, 0, 0])
    simbolos = mod_qpsk(bits_teste)

    sinal_rx_awgn = canal_awgn(simbolos, 10)
    sinal_rx_rajada = canal_rajada(simbolos, 10)

    print("Símbolos TX:      ", np.round(simbolos, 2))
    print("RX AWGN (10dB):   ", np.round(sinal_rx_awgn, 2))
    print("RX Rajada (10dB): ", np.round(sinal_rx_rajada, 2))
```

```python
import numpy as np

def calcular_ber(bits_tx, bits_rx):
    tamanho = min(len(bits_tx), len(bits_rx))
    erros = np.sum(bits_tx[:tamanho] != bits_rx[:tamanho])
    return erros / tamanho

if __name__ == "__main__":
    num_bits = 100000
    repeticoes = 5
    snr_dbs = np.arange(0, 15, 2)

    ber_bpsk = []
    ber_qpsk = []
    ber_16qam = []

    print("Iniciando a simulação do Canal AWGN...\n")

    for snr in snr_dbs:
        erros_bpsk = []
        erros_qpsk = []
        erros_16qam = []

        for _ in range(repeticoes):
            bits = np.random.randint(0, 2, num_bits)

            sym_bpsk = mod_bpsk(bits)
            sym_qpsk = mod_qpsk(bits)
            sym_16qam = mod_16qam(bits)

            rx_bpsk = canal_awgn(sym_bpsk, snr)
            rx_qpsk = canal_awgn(sym_qpsk, snr)
            rx_16qam = canal_awgn(sym_16qam, snr)

            rec_bpsk = demod_bpsk(rx_bpsk)
            rec_qpsk = demod_qpsk(rx_qpsk)
            rec_16qam = demod_16qam(rx_16qam)

            erros_bpsk.append(calcular_ber(bits, rec_bpsk))
            erros_qpsk.append(calcular_ber(bits, rec_qpsk))
            erros_16qam.append(calcular_ber(bits, rec_16qam))

        ber_bpsk.append(np.mean(erros_bpsk))
        ber_qpsk.append(np.mean(erros_qpsk))
        ber_16qam.append(np.mean(erros_16qam))

        print(f"SNR: {snr:2d} dB | BER BPSK: {ber_bpsk[-1]:.5f} | BER QPSK: {ber_qpsk[-1]:.5f} | BER 16-QAM: {ber_16qam[-1]:.5f}")

    print("\nSimulação concluída!")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.semilogy(snr_dbs, ber_bpsk, marker='o', label='BPSK', linewidth=2)
plt.semilogy(snr_dbs, ber_qpsk, marker='s', label='QPSK', linewidth=2)
plt.semilogy(snr_dbs, ber_16qam, marker='^', label='16-QAM', linewidth=2)

plt.title('Desempenho de Modulações Digitais em Canal AWGN', fontsize=14)
plt.xlabel('SNR (dB)', fontsize=12)
plt.ylabel('Taxa de Erro de Bits (BER)', fontsize=12)

plt.grid(True, which="both", ls="--", alpha=0.5)

plt.ylim([1e-6, 1])
plt.xlim([0, 14])

plt.legend(fontsize=12)

plt.tight_layout()
plt.show()

print("\n--- Tabela de Compromisso (Trade-off) na SNR de 10 dB ---\n")
print(f"{'Modulação':<12} | {'Eficiência Nominal':<20} | {'BER (SNR = 10dB)'}")
print("-" * 55)

print(f"{'BPSK':<12} | {'1 bit/s/Hz':<20} | {ber_bpsk[5]:.6f}")
print(f"{'QPSK':<12} | {'2 bits/s/Hz':<20} | {ber_qpsk[5]:.6f}")
print(f"{'16-QAM':<12} | {'4 bits/s/Hz':<20} | {ber_16qam[5]:.6f}")
```
